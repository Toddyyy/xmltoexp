# M06-2 XML 可视化网站（起步）

目标：做一个网页可视化 `Mazurka06-2.xml`，并叠加模型计算出的断点。

这个notebook会：
- 读取 `M06-2_all_levels_pred_peaks.csv`
- 读取 XML 并提取每小节音符数
- 生成 `app.py`（Streamlit 网站）


````python
from pathlib import Path

def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / 'MazurkaBL-master').exists() and (candidate / 'MERIX SUBMISSION').exists():
            return candidate
    raise FileNotFoundError('无法定位项目根目录')

PROJECT = find_project_root(Path.cwd().resolve())
XML_PATH = PROJECT / 'MazurkaBL-master/xml_scores/Mazurka06-2.xml'
PEAKS_PATH = PROJECT / 'MERIX SUBMISSION/MIREX_Model/out/M06-2_all_levels_pred_peaks.csv'
APP_PATH = PROJECT / 'MERIX SUBMISSION/M06-2_XML_WebViz/app.py'

XML_PATH, PEAKS_PATH, APP_PATH
````

````python
import re
import pandas as pd

df = pd.read_csv(PEAKS_PATH)
level = 3
values = str(df.loc[df['level'] == level, 'pred_peaks'].iloc[0])
breakpoints = [int(x) for x in re.findall(r'\d+', values)]
breakpoints[:20], len(breakpoints)
````

````python
import xml.etree.ElementTree as ET

root = ET.parse(XML_PATH).getroot()
part = root.find('part')
rows = []
for m in part.findall('measure'):
    rows.append({
        'measure': int(m.attrib.get('number', '0')),
        'note_count': len(m.findall('.//note'))
    })

measure_df = pd.DataFrame(rows)
measure_df.head(), measure_df['measure'].max()
````
